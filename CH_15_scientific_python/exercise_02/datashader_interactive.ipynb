```python
from pathlib import Path
import sys

import ipywidgets as widgets
import pandas as pd
from IPython.display import display

def find_repository_root(start: Path) -> Path:
    candidates: tuple[Path, ...] = (start, *start.parents)
    candidate: Path
    for candidate in candidates:
        if (
            (candidate / "pyproject.toml").is_file()
            and (candidate / "CH_15_scientific_python").is_dir()
        ):
            return candidate
    raise RuntimeError("repository root could not be found")

current_directory: Path = Path.cwd().resolve()
repository_root: Path = find_repository_root(current_directory)
repository_root_text: str = str(repository_root)
if repository_root != current_directory and repository_root_text not in sys.path:
    sys.path.insert(0, repository_root_text)

from CH_15_scientific_python.exercise_02.solution_00 import build_interactive_plot
```

```python
points: Path = repository_root / "CH_15_scientific_python/exercise_01/points.csv"
frame: pd.DataFrame = pd.read_csv(points)
widget: widgets.VBox = build_interactive_plot(frame)
display(widget)
```
